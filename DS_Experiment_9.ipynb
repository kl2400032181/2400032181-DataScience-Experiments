```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import bernoulli

p = 0.6

x = [0, 1]
prob = bernoulli.pmf(x, p)

print("Probabilities:", prob)

plt.bar(x, prob)
plt.xlabel("Outcome (0 = Failure, 1 = Success)")
plt.ylabel("Probability")
plt.title("Bernoulli Distribution")
plt.xticks(x)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 10
p = 0.5

x = np.arange(0, n + 1)
prob = binom.pmf(x, n, p)

print("Probabilities:", np.round(prob, 4))

plt.bar(x, prob)
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

lam = 4

x = np.arange(0, 15)
prob = poisson.pmf(x, lam)

print("Probabilities:", np.round(prob, 4))

plt.bar(x, prob)
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.title("Poisson Distribution")
plt.show()
```

```python
from scipy.stats import ttest_1samp

marks = [72, 75, 68, 70, 74, 77, 69, 73, 71, 76]
population_mean = 70

t_stat, p_value = ttest_1samp(marks, population_mean)

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

```python
from scipy.stats import ttest_ind

group_a = [70, 75, 80, 85, 78]
group_b = [65, 68, 72, 70, 67]

t_stat, p_value = ttest_ind(
    group_a, group_b, equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Significant difference between groups")
else:
    print("No significant difference between groups")
```

```python
from statsmodels.stats.weightstats import ztest

marks = [72, 75, 68, 70, 74, 77, 69, 73, 71, 76]
population_mean = 70

z_stat, p_value = ztest(
    marks, value=population_mean
)

print("Z-statistic:", z_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

```python
from scipy.stats import chi2_contingency

observed = [
    [30, 20],
    [15, 35]
]

chi2, p_value, dof, expected = chi2_contingency(observed)

print("Chi-square statistic:", chi2)
print("P-value:", p_value)
print("Degrees of freedom:", dof)
print("Expected frequencies:\n", expected)

if p_value < 0.05:
    print("Variables are significantly associated")
else:
    print("No significant association found")
```

```python
from scipy.stats import f_oneway

group_a = [60, 65, 70, 68, 62]
group_b = [70, 75, 80, 78, 72]
group_c = [80, 85, 90, 88, 82]

f_stat, p_value = f_oneway(
    group_a, group_b, group_c
)

print("F-statistic:", f_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Significant difference among group means")
else:
    print("No significant difference among group means")
```
