```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("sales_data.csv")

# Display column names
print("Columns:")
print(df.columns.tolist())

# Find a date column automatically
date_column = None

for column in df.columns:
    converted = pd.to_datetime(df[column], errors="coerce")

    if converted.notna().sum() > len(df) * 0.5:
        date_column = column
        df[column] = converted
        break

# Check whether date column was found
if date_column is None:
    print("No date column was found in the dataset.")

else:
    print("\nDate column:", date_column)

    # Create Month column
    df["Month"] = df[date_column].dt.to_period("M")

    # Find numerical columns
    numerical = df.select_dtypes(include="number").columns

    print("Numerical columns:")
    print(numerical.tolist())

    # Remove Month-related columns if any
    numerical = [
        column for column in numerical
        if column != "Month"
    ]

    # Create monthly totals
    monthly = df.groupby("Month")[numerical].sum()

    # Convert index to string
    monthly.index = monthly.index.astype(str)

    # Create line chart for each numerical column
    for column in numerical:

        plt.figure(figsize=(10, 5))

        plt.plot(
            monthly.index,
            monthly[column],
            marker="o"
        )

        plt.xlabel("Month")
        plt.ylabel(column)
        plt.title("Monthly " + column + " Trend")

        plt.xticks(rotation=45)
        plt.grid()

        plt.tight_layout()
        plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("sales_data.csv")

# Find categorical and numerical columns
categorical = df.select_dtypes(include="object").columns
numerical = df.select_dtypes(include="number").columns

print("Categorical columns:")
print(categorical.tolist())

print("\nNumerical columns:")
print(numerical.tolist())

# Check whether both types exist
if len(categorical) > 0 and len(numerical) > 0:

    # Select first categorical and first numerical column
    category = categorical[0]
    value = numerical[0]

    # Calculate total value for each category
    grouped = df.groupby(category)[value].sum()

    print("\nTotal", value, "by", category)
    print(grouped)

    # Create bar chart
    plt.figure(figsize=(8, 5))

    plt.bar(grouped.index.astype(str), grouped.values)

    plt.xlabel(category)
    plt.ylabel("Total " + value)
    plt.title("Total " + value + " by " + category)

    plt.xticks(rotation=45)
    plt.tight_layout()

    plt.show()

else:
    print("Categorical or numerical columns are missing.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("sales_data.csv")

# Find categorical and numerical columns
categorical = df.select_dtypes(include="object").columns
numerical = df.select_dtypes(include="number").columns

print("Categorical columns:")
print(categorical.tolist())

print("\nNumerical columns:")
print(numerical.tolist())

# We need at least 2 categorical columns
# and at least 1 numerical column
if len(categorical) >= 2 and len(numerical) >= 1:

    category1 = categorical[0]
    category2 = categorical[1]
    value = numerical[0]

    # Create pivot table
    pivot = pd.pivot_table(
        df,
        values=value,
        index=category1,
        columns=category2,
        aggfunc="sum",
        fill_value=0
    )

    print("\nPivot Table:")
    print(pivot)

    # Create grouped bar chart
    pivot.plot(
        kind="bar",
        figsize=(10, 6)
    )

    plt.xlabel(category1)
    plt.ylabel("Total " + value)
    plt.title(
        value + " by " + category1 +
        " and " + category2
    )

    plt.xticks(rotation=45)
    plt.legend(title=category2)

    plt.tight_layout()
    plt.show()

else:
    print(
        "The dataset must contain at least "
        "2 categorical and 1 numerical column."
    )
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("sales_data.csv")

# Select numerical columns
numerical = df.select_dtypes(include="number")

print("Numerical Attributes:")
print(numerical.columns.tolist())

# Calculate correlation matrix
correlation = numerical.corr()

print("\nCorrelation Matrix:")
print(correlation)

# Create heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap")
plt.tight_layout()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# -------------------------------------------------
# 1. Load Dataset
# -------------------------------------------------

df = pd.read_csv("sales_data.csv")

print("Dataset loaded successfully!")

# -------------------------------------------------
# 2. Identify Columns
# -------------------------------------------------

numerical = df.select_dtypes(include="number").columns.tolist()
categorical = df.select_dtypes(include="object").columns.tolist()

print("\nNumerical Columns:")
print(numerical)

print("\nCategorical Columns:")
print(categorical)

# -------------------------------------------------
# 3. Find Date Column
# -------------------------------------------------

date_column = None

for column in df.columns:

    converted = pd.to_datetime(
        df[column],
        errors="coerce"
    )

    if converted.notna().sum() > len(df) * 0.5:

        date_column = column
        df[column] = converted
        break

# -------------------------------------------------
# 4. Create Dashboard
# -------------------------------------------------

fig = plt.figure(figsize=(16, 12))

# =================================================
# 5. Histogram
# =================================================

ax1 = plt.subplot(3, 3, 1)

if len(numerical) > 0:

    column = numerical[0]

    ax1.hist(
        df[column].dropna(),
        bins=10,
        edgecolor="black"
    )

    ax1.set_title("Histogram - " + column)
    ax1.set_xlabel(column)
    ax1.set_ylabel("Frequency")

else:

    ax1.text(
        0.5,
        0.5,
        "No numerical data",
        ha="center"
    )

# =================================================
# 6. Boxplot
# =================================================

ax2 = plt.subplot(3, 3, 2)

if len(numerical) > 0:

    column = numerical[0]

    ax2.boxplot(
        df[column].dropna()
    )

    ax2.set_title("Boxplot - " + column)
    ax2.set_ylabel(column)

else:

    ax2.text(
        0.5,
        0.5,
        "No numerical data",
        ha="center"
    )

# =================================================
# 7. Scatterplot
# =================================================

ax3 = plt.subplot(3, 3, 3)

if len(numerical) >= 2:

    x = numerical[0]
    y = numerical[1]

    ax3.scatter(
        df[x],
        df[y]
    )

    ax3.set_title(
        x + " vs " + y
    )

    ax3.set_xlabel(x)
    ax3.set_ylabel(y)

else:

    ax3.text(
        0.5,
        0.5,
        "Need 2 numerical columns",
        ha="center"
    )

# =================================================
# 8. Line Chart
# =================================================

ax4 = plt.subplot(3, 3, 4)

if date_column is not None and len(numerical) > 0:

    value = numerical[0]

    df["Month"] = df[date_column].dt.to_period("M")

    monthly = df.groupby("Month")[value].sum()

    ax4.plot(
        monthly.index.astype(str),
        monthly.values,
        marker="o"
    )

    ax4.set_title(
        "Monthly " + value + " Trend"
    )

    ax4.set_xlabel("Month")
    ax4.set_ylabel(value)

    ax4.tick_params(
        axis="x",
        rotation=45
    )

else:

    ax4.text(
        0.5,
        0.5,
        "Date data not available",
        ha="center"
    )

# =================================================
# 9. Bar Chart
# =================================================

ax5 = plt.subplot(3, 3, 5)

if len(categorical) > 0 and len(numerical) > 0:

    category = categorical[0]
    value = numerical[0]

    grouped = df.groupby(category)[value].sum()

    ax5.bar(
        grouped.index.astype(str),
        grouped.values
    )

    ax5.set_title(
        value + " by " + category
    )

    ax5.set_xlabel(category)
    ax5.set_ylabel(value)

    ax5.tick_params(
        axis="x",
        rotation=45
    )

else:

    ax5.text(
        0.5,
        0.5,
        "Required data not available",
        ha="center"
    )

# =================================================
# 10. Second Line Chart
# =================================================

ax6 = plt.subplot(3, 3, 6)

if date_column is not None and len(numerical) >= 2:

    value = numerical[1]

    monthly = df.groupby("Month")[value].sum()

    ax6.plot(
        monthly.index.astype(str),
        monthly.values,
        marker="o"
    )

    ax6.set_title(
        "Monthly " + value + " Trend"
    )

    ax6.set_xlabel("Month")
    ax6.set_ylabel(value)

    ax6.tick_params(
        axis="x",
        rotation=45
    )

else:

    ax6.text(
        0.5,
        0.5,
        "Second numerical trend unavailable",
        ha="center"
    )

# =================================================
# 11. Product/Category Bar Chart
# =================================================

ax7 = plt.subplot(3, 3, 7)

if len(categorical) >= 2 and len(numerical) > 0:

    category = categorical[1]
    value = numerical[0]

    grouped = df.groupby(category)[value].sum()

    ax7.bar(
        grouped.index.astype(str),
        grouped.values
    )

    ax7.set_title(
        value + " by " + category
    )

    ax7.set_xlabel(category)
    ax7.set_ylabel(value)

    ax7.tick_params(
        axis="x",
        rotation=45
    )

else:

    ax7.text(
        0.5,
        0.5,
        "Second category unavailable",
        ha="center"
    )

# =================================================
# 12. Correlation Heatmap
# =================================================

ax8 = plt.subplot(3, 3, 8)

if len(numerical) >= 2:

    correlation = df[numerical].corr()

    sns.heatmap(
        correlation,
        annot=True,
        fmt=".2f",
        cmap="coolwarm",
        ax=ax8
    )

    ax8.set_title(
        "Correlation Heatmap"
    )

else:

    ax8.text(
        0.5,
        0.5,
        "Need numerical columns",
        ha="center"
    )

# =================================================
# 13. Pivot Table
# =================================================

ax9 = plt.subplot(3, 3, 9)

if len(categorical) >= 2 and len(numerical) > 0:

    category1 = categorical[0]
    category2 = categorical[1]
    value = numerical[0]

    pivot = pd.pivot_table(
        df,
        values=value,
        index=category1,
        columns=category2,
        aggfunc="sum",
        fill_value=0
    )

    print("\nPivot Table:")
    print(pivot)

    ax9.axis("off")

    table = ax9.table(
        cellText=pivot.round(2).values,
        rowLabels=pivot.index,
        colLabels=pivot.columns,
        loc="center"
    )

    table.auto_set_font_size(False)
    table.set_fontsize(7)

    ax9.set_title(
        "Pivot Table"
    )

else:

    ax9.text(
        0.5,
        0.5,
        "Pivot table unavailable",
        ha="center"
    )

# -------------------------------------------------
# 14. Display Dashboard
# -------------------------------------------------

plt.suptitle(
    "Exploratory Data Analysis Dashboard",
    fontsize=18
)

plt.tight_layout()

plt.show()
```
