```python
import numpy as np

# Create two vectors
A = np.array([1, 2, 3])
B = np.array([4, 5, 6])

# Vector addition
addition = A + B

# Vector subtraction
subtraction = A - B

# Scalar multiplication
scalar = 3
scalar_multiplication = scalar * A

# Dot product
dot_product = np.dot(A, B)

# Magnitude of vectors
magnitude_A = np.linalg.norm(A)
magnitude_B = np.linalg.norm(B)

# Display results
print("Vector A:", A)
print("Vector B:", B)

print("\nVector Addition:", addition)
print("Vector Subtraction:", subtraction)
print("Scalar Multiplication:", scalar_multiplication)
print("Dot Product:", dot_product)

print("Magnitude of Vector A:", magnitude_A)
print("Magnitude of Vector B:", magnitude_B)
```

```python
import numpy as np

# Create a square matrix
A = np.array([
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 10]
])

# Calculate determinant
determinant = np.linalg.det(A)

# Calculate rank
rank = np.linalg.matrix_rank(A)

# Calculate trace
trace = np.trace(A)

# Display matrix
print("Matrix A:")
print(A)

# Display results
print("\nDeterminant:", determinant)
print("Rank:", rank)
print("Trace:", trace)
```

```python
import numpy as np

# Create a square matrix
A = np.array([
    [4, 1],
    [2, 3]
])

# Calculate eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)

# Display matrix
print("Matrix A:")
print(A)

# Display eigenvalues
print("\nEigenvalues:")
print(eigenvalues)

# Display eigenvectors
print("\nEigenvectors:")
print(eigenvectors)
```

```python
import numpy as np

# Create a matrix
A = np.array([
    [4, 1],
    [2, 3]
])

# Calculate eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)

# Select the first eigenvalue and corresponding eigenvector
lambda_value = eigenvalues[0]
v = eigenvectors[:, 0]

# Calculate Av
Av = A @ v

# Calculate lambda * v
lambda_v = lambda_value * v

# Display results
print("Matrix A:")
print(A)

print("\nEigenvalue:", lambda_value)
print("\nEigenvector:", v)

print("\nA @ v:")
print(Av)

print("\nLambda * v:")
print(lambda_v)

# Verify the relationship
if np.allclose(Av, lambda_v):
    print("\nAv = λv is verified.")
else:
    print("\nAv = λv is not verified.")
```

```python
import numpy as np
from scipy.linalg import lu

# Create a square matrix
A = np.array([
    [4, 3],
    [6, 3]
])

# Perform LU decomposition
P, L, U = lu(A)

# Display results
print("Matrix A:")
print(A)

print("\nPermutation Matrix P:")
print(P)

print("\nLower Triangular Matrix L:")
print(L)

print("\nUpper Triangular Matrix U:")
print(U)
```

```python
import numpy as np

# Create a symmetric positive definite matrix
A = np.array([
    [4, 2],
    [2, 3]
])

# Perform Cholesky decomposition
L = np.linalg.cholesky(A)

# Display matrix
print("Matrix A:")
print(A)

# Display lower triangular matrix
print("\nLower Triangular Matrix L:")
print(L)

# Verify A = L @ L.T
print("\nL @ L.T:")
print(L @ L.T)
```

```python
import numpy as np

# Create a matrix
A = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

# Perform QR decomposition
Q, R = np.linalg.qr(A)

# Display matrix
print("Matrix A:")
print(A)

# Display Q
print("\nOrthogonal Matrix Q:")
print(Q)

# Display R
print("\nUpper Triangular Matrix R:")
print(R)

# Verify A = Q @ R
print("\nQ @ R:")
print(Q @ R)
```

```python
import numpy as np

# Create a matrix
A = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

# Perform SVD
U, S, VT = np.linalg.svd(A)

# Create Sigma matrix
Sigma = np.zeros((U.shape[1], VT.shape[0]))
np.fill_diagonal(Sigma, S)

# Display original matrix
print("Matrix A:")
print(A)

# Display U
print("\nMatrix U:")
print(U)

# Display singular values
print("\nSingular Values:")
print(S)

# Display Sigma
print("\nSigma Matrix:")
print(Sigma)

# Display V transpose
print("\nMatrix V^T:")
print(VT)

# Verify A = U @ Sigma @ VT
print("\nReconstructed Matrix:")
print(U @ Sigma @ VT)
```
